### Eksekusi 1: Geolistrik Resistivitas
Data sintetis: lintasan pengukuran sepanjang 120 m dengan kedalaman model 25 m. Model terdiri atas tiga zona resistivitas yang berbeda untuk menggambarkan variasi kondisi bawah permukaan.

Kode dan visualisasi berikut menggunakan data sintetis untuk demonstrasi pemodelan resistivitas 2D.

Tujuan eksekusi: menghasilkan model dan visualisasi penampang resistivitas 2D untuk mengidentifikasi persebaran nilai resistivitas terhadap jarak dan kedalaman bawah permukaan.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Pembuatan model resistivitas sintetis
x = np.linspace(0, 120, 121)
z = np.linspace(0, 25, 51)
X, Z = np.meshgrid(x, z)

rho = np.full_like(X, 80.0)
rho[(X > 25) & (X < 55) & (Z > 5) & (Z < 17)] = 22
rho[(X > 78) & (Z > 9)] = 160

# 2. Informasi model
print("METODE GEOLISTRIK RESISTIVITAS")
print(f"Resistivitas minimum: {rho.min():.1f} ohm.m")
print(f"Resistivitas maksimum: {rho.max():.1f} ohm.m")
print("Panjang lintasan: 120 m")
print("Kedalaman model: 25 m")

# 3. Visualisasi penampang resistivitas
plt.figure(figsize=(9, 4.5))
plt.pcolormesh(X, Z, rho, shading="auto", cmap="viridis")
plt.gca().invert_yaxis()
plt.xlabel("Jarak lintasan (m)")
plt.ylabel("Kedalaman (m)")
plt.title("Model Resistivitas Bawah Permukaan")
plt.colorbar(label="Resistivitas (ohm.m)")
plt.tight_layout()
plt.show()
```

Model menunjukkan resistivitas minimum 22 Ωm dan maksimum 160 Ωm. Perbedaan nilai tersebut menggambarkan tiga zona resistivitas sintetis yang dapat diamati berdasarkan posisi dan kedalamannya.

### Eksekusi 2: Seismik Refraksi
Data sintetis: jarak pengukuran 0–120 m dengan dua lapisan model yang memiliki kecepatan gelombang masing-masing 450 m/s dan 1.600 m/s.

Kode dan visualisasi berikut menggunakan data sintetis untuk memperlihatkan hubungan antara jarak dan waktu tempuh gelombang seismik.

Tujuan eksekusi: menghasilkan kurva waktu tempuh gelombang untuk memperlihatkan perbedaan kecepatan rambat gelombang pada model dua lapisan bawah permukaan.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Pembuatan data waktu tempuh sintetis
jarak = np.linspace(0, 120, 121)
v1 = 450.0
v2 = 1600.0

waktu_langsung = jarak / v1
waktu_refraksi = 0.025 + jarak / v2
waktu_tiba = np.minimum(waktu_langsung, waktu_refraksi)

# 2. Informasi model
print("METODE SEISMIK REFRAKSI")
print(f"Kecepatan lapisan 1: {v1:.0f} m/s")
print(f"Kecepatan lapisan 2: {v2:.0f} m/s")
print(f"Waktu tiba pertama pada 120 m: {waktu_tiba[-1]*1000:.2f} ms")

# 3. Visualisasi kurva waktu tempuh
plt.figure(figsize=(8.5, 4.5))
plt.plot(jarak, waktu_langsung*1000, label="Gelombang langsung")
plt.plot(jarak, waktu_refraksi*1000, label="Gelombang refraksi")
plt.plot(jarak, waktu_tiba*1000, "--", label="Waktu tiba pertama")
plt.xlabel("Jarak sumber-penerima (m)")
plt.ylabel("Waktu tempuh (ms)")
plt.title("Kurva Waktu Tempuh Seismik Refraksi")
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

Grafik menunjukkan perubahan waktu tempuh terhadap jarak. Perbedaan kemiringan kurva menggambarkan perbedaan kecepatan gelombang pada model. Data ini belum digunakan untuk menentukan kedalaman lapisan berdasarkan hasil survei nyata.

### Eksekusi 3: Seismik Refleksi
Data sintetis: penampang seismik dengan rentang jarak 0–1.000 m dan waktu dua arah 0–1,5 detik. Model terdiri atas empat reflektor dengan geometri yang bervariasi.

Kode dan visualisasi berikut menggunakan data sintetis untuk mendemonstrasikan tampilan penampang refleksi seismik.

Tujuan eksekusi: menghasilkan penampang seismik sintetis untuk mengamati pola reflektor, perubahan geometri perlapisan, dan karakteristik gelombang pantul terhadap waktu.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Pembuatan penampang seismik sintetis
rng = np.random.default_rng(42)

x = np.linspace(0, 1000, 220)
t = np.linspace(0, 1.5, 250)
data = 0.10 * rng.normal(size=(len(t), len(x)))

reflektor = [
    0.30 + 0.025*np.sin(x/130),
    0.62 + 0.040*np.sin(x/170),
    0.92 + 0.00015*(x-500),
    1.20 + 0.025*np.cos(x/150)
]

for i, waktu in enumerate(reflektor):
    indeks = np.round(waktu/(t[1]-t[0])).astype(int)
    for j, k in enumerate(indeks):
        data[k-1:k+2, j] += (-1)**i * np.array([0.4, 1.0, -0.5])

# 2. Informasi model
print("METODE SEISMIK REFLEKSI")
print(f"Jumlah reflektor sintetis: {len(reflektor)}")
print(f"Rentang waktu: {t.min():.2f}–{t.max():.2f} s")
print(f"Jumlah titik horizontal: {len(x)}")

# 3. Visualisasi penampang refleksi
batas = np.percentile(np.abs(data), 98)

plt.figure(figsize=(9, 4.5))
plt.imshow(
    data,
    extent=[x.min(), x.max(), t.max(), t.min()],
    aspect="auto",
    cmap="gray",
    vmin=-batas,
    vmax=batas
)
plt.xlabel("Jarak lintasan (m)")
plt.ylabel("Waktu dua arah (s)")
plt.title("Penampang Seismik Refleksi Sintetis")
plt.tight_layout()
plt.show()
```

Penampang menampilkan empat reflektor sintetis dengan bentuk bergelombang dan perubahan posisi. Pola tersebut digunakan untuk latihan pengamatan geometri perlapisan, bukan sebagai bukti langsung adanya struktur geologi tertentu.

### Eksekusi 4: Geomagnetik
Data sintetis: area model berukuran 1.200 × 900 m dengan dua sumber anomali magnetik yang menghasilkan respons positif dan negatif.

Kode dan visualisasi berikut menggunakan data sintetis untuk memperlihatkan distribusi anomali magnetik secara spasial.

Tujuan eksekusi: menghasilkan peta kontur anomali magnetik untuk mengamati distribusi anomali positif dan negatif serta memahami variasi respons magnetik bawah permukaan.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Pembuatan model anomali magnetik
x = np.linspace(0, 1200, 121)
y = np.linspace(0, 900, 91)
X, Y = np.meshgrid(x, y)

anomali = (
    75*np.exp(-(((X-350)/180)**2 + ((Y-500)/160)**2))
    - 55*np.exp(-(((X-850)/220)**2 + ((Y-300)/180)**2))
)

# 2. Informasi model
print("METODE GEOMAGNETIK")
print(f"Anomali minimum: {anomali.min():.2f} nT")
print(f"Anomali maksimum: {anomali.max():.2f} nT")
print("Cakupan model: 1200 x 900 m")

# 3. Visualisasi peta anomali magnetik
plt.figure(figsize=(8.5, 4.8))
kontur = plt.contourf(X, Y, anomali, levels=20, cmap="RdBu_r")
plt.contour(X, Y, anomali, levels=10, colors="black", linewidths=0.4)
plt.xlabel("Koordinat X (m)")
plt.ylabel("Koordinat Y (m)")
plt.title("Peta Anomali Magnetik Sintetis")
plt.colorbar(kontur, label="Anomali magnetik (nT)")
plt.tight_layout()
plt.show()
```

Peta memperlihatkan distribusi anomali positif dan negatif. Perbedaan respons magnetik dapat berkaitan dengan variasi sifat kemagnetan material bawah permukaan, tetapi penentuan litologi membutuhkan informasi pendukung.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Pembuatan model anomali gravitasi
x = np.linspace(0, 1500, 151)
y = np.linspace(0, 1000, 101)
X, Y = np.meshgrid(x, y)

anomali = (
    4.2*np.exp(-(((X-450)/250)**2 + ((Y-650)/210)**2))
    - 2.8*np.exp(-(((X-1100)/300)**2 + ((Y-350)/230)**2))
)

# 2. Informasi model
print("METODE GRAVITASI")
print(f"Anomali minimum: {anomali.min():.2f} mGal")
print(f"Anomali maksimum: {anomali.max():.2f} mGal")
print("Cakupan model: 1500 x 1000 m")

# 3. Visualisasi peta anomali gravitasi
plt.figure(figsize=(8.5, 4.8))
kontur = plt.contourf(X, Y, anomali, levels=20, cmap="Spectral_r")
plt.contour(X, Y, anomali, levels=10, colors="black", linewidths=0.4)
plt.xlabel("Koordinat X (m)")
plt.ylabel("Koordinat Y (m)")
plt.title("Peta Anomali Gravitasi Sintetis")
plt.colorbar(kontur, label="Anomali gravitasi (mGal)")
plt.tight_layout()
plt.show()
```
